## Task 2: Run all of Python commands in the table "Linear Algebra Equivalents" in Numpy for MATLAB Users.

```python
import numpy as np
from scipy import linalg
from scipy.sparse.linalg import eigs, cg
from scipy import signal
from IPython.display import display

a = np.array([
    [4.0, 1.0, 0.5, 0.2],
    [1.0, 3.0, 0.7, 0.3],
    [0.5, 0.7, 2.0, 0.4],
    [0.2, 0.3, 0.4, 1.0]
])

b = np.eye(4) * 2.0
b_vec = np.ones(4)

Z = np.random.rand(10, 4)
y = np.random.rand(10)
x_sig = np.arange(100.0)
q = 4


print("linalg.inv(a):")
display(linalg.inv(a))

print("linalg.pinv(a):")
display(linalg.pinv(a))

print("np.linalg.matrix_rank(a):")
display(np.linalg.matrix_rank(a))

print("linalg.solve(a, b):")
display(linalg.solve(a, b))

print("linalg.lstsq(a, b):")
display(linalg.lstsq(a, b))

print("x = linalg.solve(a.T, b.T).T:")
x = linalg.solve(a.T, b.T).T
display(x)

print("SVD: U, S, Vh = linalg.svd(a)")
U, S, Vh = linalg.svd(a)
print("U:"); display(U)
print("S:"); display(S)
print("Vh:"); display(Vh)
V = Vh.T
print("V = Vh.T:"); display(V)

print("linalg.cholesky(a):")
display(linalg.cholesky(a))

print("Eigen decomposition: D, V = linalg.eig(a)")
D, V = linalg.eig(a)
print("D:"); display(D)
print("V:"); display(V)

print("Generalized eigen: D, V = linalg.eig(a, b)")
D, V = linalg.eig(a, b)
print("D:"); display(D)
print("V:"); display(V)

print("eigs(a, k=2): D, V = eigs(a, k=2)")
D, V = eigs(a, k=2)
print("D:"); display(D)
print("V:"); display(V)

print("QR: Q, R = linalg.qr(a)")
Q, R = linalg.qr(a)
print("Q:"); display(Q)
print("R:"); display(R)

print("LU: P, L, U = linalg.lu(a)")
P, L, U = linalg.lu(a)
print("P:"); display(P)
print("L:"); display(L)
print("U:"); display(U)

print("Conjugate gradients: x_cg, info = cg(a, b_vec)")
x_cg, info = cg(a, b_vec)
print("x_cg:"); display(x_cg)
print("info:", info)

print("FFT: np.fft.fft(a)")
display(np.fft.fft(a))

print("IFFT: np.fft.ifft(a)")
display(np.fft.ifft(a))

print("np.sort(a):")
display(np.sort(a))

print("a_copy.sort(axis=0):")
a_copy = a.copy()
a_copy.sort(axis=0)
display(a_copy)

print("np.sort(a, axis=1):")
display(np.sort(a, axis=1))

print("a_copy.sort(axis=1):")
a_copy = a.copy()
a_copy.sort(axis=1)
display(a_copy)

print("Sort rows by first column:")
I = np.argsort(a[:, 0])
b_sorted = a[I, :]
print("I:"); display(I)
print("b_sorted:"); display(b_sorted)

print("Linear regression: x_lstsq = linalg.lstsq(Z, y)[0]")
x_lstsq = linalg.lstsq(Z, y)[0]
display(x_lstsq)

print("Downsample: signal.resample(x_sig, ceil(len(x_sig)/q))")
display(signal.resample(x_sig, int(np.ceil(len(x_sig) / q))))

print("np.unique(a):")
display(np.unique(a))

print("a.squeeze():")
display(a.squeeze())
```
